### 1.3 Create a box plot

```python
# Create data for cell 0
df_cells_with_zero = df[df["Cell"] == 0]
df_filtered_data = df_cells_with_zero.loc[:, ['Notebook', 'Value']]
```

```python
# Create data for cell 2 to the end
df_non_zero_cells = df.loc[df['Cell'] != 0]
```

```python
df_median_per_notebook = df_non_zero_cells.groupby('Notebook', as_index=False)['Value'].median()
```

```python
# Set the size of the figure
plt.figure(figsize=(12, 6))

# Define box and median properties for both subplots
boxprops_cell0 = dict(linewidth=2, color='blue')
medianprops_cell0 = dict(linewidth=2, color='red')
boxprops_merged_cell = dict(linewidth=2, color='green')
medianprops_merged_cell = dict(linewidth=2, color='purple')
```

```python
# First boxplot for "cell0"
plt.subplot(1, 2, 1)
plt.boxplot(df_median_per_notebook['Value'], boxprops=boxprops_cell0, medianprops=medianprops_cell0)
plt.title('Boxplot for cell0')
plt.xticks([1], ['cell0'])

# Add annotations for statistics
statistics_cell0 = df_median_per_notebook['Value'].describe()
offsets_cell0 = [0.5,2,2.5, 0, 0]
for i, stat_index in enumerate(['25%', '50%', '75%', 'max', 'min']):
    plt.annotate(f'{stat_index}: {statistics_cell0[stat_index]:.2f}',
                 xy=(1, statistics_cell0[stat_index]),
                 xytext=(1.2, statistics_cell0[stat_index] + offsets_cell0[i]),
                 color='red')

# Adjust the layout for a better appearance
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Second boxplot for "merged_cell"
plt.subplot(1, 2, 2)
plt.boxplot(df_median_per_notebook['Value'], boxprops=boxprops_merged_cell, medianprops=medianprops_merged_cell)
plt.title('Boxplot for merged_cell')
plt.xticks([1], ['merged_cell'])

# Add annotations for statistics
statistics_merged_cell = df_median_per_notebook['Value'].describe()
offsets_merged_cell = [0, 0, 0, 0, 0]
for i, stat_index in enumerate(['25%', '50%', '75%', 'max', 'min']):
    plt.annotate(f'{stat_index}: {statistics_merged_cell[stat_index]:.2f}',
                 xy=(1, statistics_merged_cell[stat_index]),
                 xytext=(1.2, statistics_merged_cell[stat_index] + offsets_merged_cell[i]),
                 color='purple')

# Adjust the layout for a better appearance
plt.tight_layout()

# Show the plot
plt.show()
```
