```python
ftable_full=np.logspace(0,np.log10(2),13)*262
ftable = ftable_full[[0,2,4,5,7,9,11]]
```

```python
ftable
```

```python
x = np.arange(0, .5, 1/rate)
for s in [5,3,3,4,2,2,1,2,3,4,5,5,5]:
    display(Audio(np.sin(x*2*np.pi*ftable[s-1]), 
                  rate=rate, autoplay=True))
    time.sleep(0.5)
```

```python
x = np.arange(0, .5, 1/rate)
song = []
for s in [5,3,3,4,2,2,1,2,3,4,5,5,5]:
    y = np.sin(x*2*np.pi*ftable[s-1])
    song.append(y)
```

```python
Audio(data=np.array(song).ravel(), rate=rate)
```

```python
Audio(data=np.concatenate(song), rate=rate)
```

```python
all_song = np.array(song).ravel()
```

```python
Audio(data=(all_song[480*5:] + all_song[:-480*5])/2, rate=rate)
```
