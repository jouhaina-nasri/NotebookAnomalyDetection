```python
import pickle
import os
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from ANEMO import read_edf
from ANEMO import ANEMO
```

```python
sujet = ['AM','BMC','CS','DC','FM','IP','LB','OP','RS','SR','TN','YK']
time = ['2017-10-23_100057','2017-09-26_095637', '2017-10-03_134421','2017-09-27_161040',
        '2017-10-03_143803','2017-09-28_115250', '2017-09-20_151043','2017-10-26_121823',
        '2017-11-08_094717','2017-11-16_153313', '2017-11-08_150410','2017-11-17_172706']
```

```python
def regress(ax, x, y, t_label=20, color='r') :
    from scipy import stats
    x1, x2, y1, y2 = min(x), max(x), min(y), max(y)
    
    slope, intercept, r_, p_value, std_err = stats.linregress(x, y)
    x_test = np.linspace(x1, x2, 100)
    fitLine = slope * x_test + intercept
    ax.plot(x_test, fitLine, c=color, linewidth=2)
    ax.text(x2-(x2-x1)/10,y1+(y2-y1)/10, 'r = %0.3f'%(r_), color=color, fontsize=t_label/1.2, ha='right')
    return ax

def scat(ax, X_, Y_, param, xname, yname, name_sujet, param_goodness_of_fit) :
    ax.scatter(X_, Y_, c='k', s=10, alpha=0.4)
    ax = regress(ax, X_, Y_, t_label=20, color='r')
    ax.set_xlabel(xname) ; ax.set_ylabel(yname)

    if param == 'a_anti' :                title = 'a_anti/time_anti'
    elif param == 'start_anti' :          title = 'start_anti for |a_anti| > 10°/s²'
    elif param == 'dist_lat_first_sacc' : title = 'Distance entre latency et première saccade'
    else :                                title = param          

    if param_goodness_of_fit is True : ax.set_xscale('log') ; ax.set_yscale('log')

    ax.set_title(name_sujet+' -- '+title+'\n %s : %0.1f - %s : %0.1f'%(xname, np.mean(X_), yname, np.mean(Y_)))
    
    return ax
```

```python
def figure(param, list_fct=None, Title=None, param_goodness_of_fit=False,  plot='hist') :
    
    if plot=='hist' :
        list_fct = ['velocity_exp', 'velocity_sigmo', 'velocity_line']
    
    fig_size = 7
    fig, ax = plt.subplots(1, 1, figsize=(fig_size, (fig_size/1.6180)))
    
    fig0, ax0 = plt.subplots(int(len(sujet)/3), int(len(sujet)/4), figsize=(fig_size*(len(sujet)/4), (len(sujet)/3)*(fig_size/1.6180)))
    
    if param == 'a_anti' :                title = 'a_anti/time_anti'
    elif param == 'start_anti' :          title = 'start_anti for |a_anti| > 10°/s²'
    elif param == 'dist_lat_first_sacc' : title = 'Distance entre latency et première saccade'
    else :                                title = param          

    if plot=='hist' :
        fig.suptitle(title, fontsize=20, y=1.1)
    elif plot=='scat':
        fig.suptitle(Title, fontsize=20, y=1.1)
    
    c = ['k', 'r', 'c']
    p_full = {}
    for fct in list_fct :
        p_full[fct] = []
    
    a, b = 0, 0
    for s in range(len(sujet)) :
        p_, p = {}, {}
        for fct in list_fct :
            if fct == 'velocity_exp' : name_file = 'velocity'
            else : name_file = fct
            
            with open('parametre/fct_%s_filt_velo_30/param_Fit_%s_fct_%s_filt_velo_30.pkl'%(name_file, sujet[s], name_file), 'rb') as fichier :
                p_[fct] = pickle.load(fichier, encoding='latin1')
        
            if param_goodness_of_fit is True :
                p_[fct] = p_[fct]['goodness_of_fit']  
            p[fct] = []
        
        #-------------------------------------------------------------------------
        if param == 'dist_lat_first_sacc' :
            file = os.path.join('data', 'enregistrement_' + sujet[s] + '_' + time[s] + '.pkl')
            with open(file, 'rb') as fichier : param_exp = pickle.load(fichier, encoding='latin1')
            A = ANEMO(param_exp)
            N_trials, N_blocks = param_exp['N_trials'], param_exp['N_blocks']

            resultats = os.path.join('data', 'enregistrement_' + sujet[s] + '_' + time[s] + '.asc')
            data = read_edf(resultats, 'TRIALID')
            
            latence = {}
            for fct in list_fct :
                latence[fct] = []
                        
            lat_sacc_1 = []
            for block in range(3) :
                for trial in range(N_trials) :
                    arg = A.arg(data[trial + N_trials*block], trial=trial, block=block)

                    lat_firt_sacc = []
                    for s in range(len(arg.saccades)) :
                        sacc = arg.saccades[s] 
                        if sacc[0] >= (arg.TargetOn+100) : lat_firt_sacc.append((sacc[0]-arg.TargetOn))
                    if lat_firt_sacc == [] :               lat_firt_sacc.append(np.nan)
                    lat_firt_sacc = lat_firt_sacc[0]
                    lat_sacc_1.append(lat_firt_sacc)
                
                for fct in list_fct :
                    latence[fct].extend(p_[fct]['latency'][block])
            
            for fct in list_fct :
                latence[fct] = np.ma.masked_array(latence[fct], mask=np.isnan(lat_sacc_1)).compressed()
            lat_sacc_1 = np.ma.masked_array(lat_sacc_1, mask=np.isnan(lat_sacc_1)).compressed()

            for fct in list_fct :
                p[fct] = list(np.array(lat_sacc_1)-np.array(latence[fct]))
    
        #-------------------------------------------------------------------------
        else :
            for block in range(3) :
                for fct in list_fct :
                    if param == 'a_anti' :
                        p[fct].extend(np.array(p_[fct][param][block])*((np.array(p_[fct]['latency'][block])-np.array(p_[fct]['start_anti'][block]))/1000))

                    elif param == 'start_anti' :
                        sa = []
                        for t in range(len(p_[fct]['start_anti'][block])) :
                            if plot=='hist' :
                                if abs(p_[fct]['a_anti'][block][t]) > 10 :
                                    sa.append(p_[fct]['start_anti'][block][t])
                            elif plot=='scat' :
                                 if (abs(p_[list_fct[0]]['a_anti'][block][t]) and abs(p_[list_fct[1]]['a_anti'][block][t])) > 10 :
                                    sa.append(p_[fct]['start_anti'][block][t])
                        p[fct].extend(sa)
                    elif param=='ramp' :
                        if fct == 'velocity_exp' : par = 'tau'
                        else :                     par = 'ramp_pursuit'
                        p[fct].extend(p_[fct][par][block])
                    else :
                        p[fct].extend(p_[fct][param][block])
        
        #-------------------------------------------------------------------------
        for fct in list_fct :
            p_full[fct].extend(p[fct])
            
        if plot=='hist' :
            for i, fct in enumerate(list_fct) :
                ax0[a][b].hist(p[fct], bins=40, color=c[i], alpha=0.5, label='%s -- %0.3f'%(fct, np.mean(p[fct])))
            ax0[a][b].set_title(sujet[s]+' -- '+title)
            ax0[a][b].legend(fontsize=12)
        
        elif plot=='scat' :
            ax0[a][b] = scat(ax0[a][b], p[list_fct[0]], p[list_fct[1]], param,
                             list_fct[0], list_fct[1], sujet[s], param_goodness_of_fit)
        
        b+=1
        if b==3 : b=0 ; a+=1
    
    #-------------------------------------------------------------------------
    if plot=='hist' :
        for i, fct in enumerate(list_fct) :
            ax.hist(p_full[fct], bins=40, color=c[i], alpha=0.5, label='%s -- %0.3f'%(fct, np.mean(p_full[fct])))
        ax.set_title('Full sujet -- '+title)
        ax.legend(fontsize=12)
    elif plot=='scat' :
        ax = scat(ax, p_full[list_fct[0]], p_full[list_fct[1]], param,
                 list_fct[0], list_fct[1], 'Full sujet', param_goodness_of_fit)
    
    plt.subplots_adjust(hspace=0.5)
    plt.show()
```
