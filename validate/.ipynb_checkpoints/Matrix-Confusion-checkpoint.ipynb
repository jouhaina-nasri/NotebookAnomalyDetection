# Validation 2 : Matrix Confusion

## 1- Imports and Setup

```python
import pandas as pd
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2- Load Data

Here, we load the data from the provided CSV files.

```python
tool_df = pd.read_csv('result_tool.csv')
manual_df = pd.read_csv('manual_labeling.csv')
```

## 3- Define Labels List

List of specific labels or code smells we're interested in.

```python
code_smells_list = [
    'LC', 
    'LEC', 
    'CCC', 
    'LE', 
    'CIC'
]
```

## 4- Generate and Display Confusion Matrices

We iterate over each label (code smell) and generate confusion matrices. These are then visualized using Seaborn's heatmap.

```python
for smell in code_smells_list:
    tool_labels = tool_df[smell]
    manual_labels = manual_df[smell]

    # Compute the confusion matrix
    cm = confusion_matrix(manual_labels, tool_labels)

    # Display the confusion matrix
    plt.figure(figsize=(10, 7))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f"Confusion Matrix for {smell}")
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```
