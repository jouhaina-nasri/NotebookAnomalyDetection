# Comparing DataFrames ( Manuel Labeling + Result )

## 1- Loading the Data

First, we will load the data from the CSV files into DataFrames.

```python
import pandas as pd
```

```python
df1 = pd.read_csv('result.csv')
df2 = pd.read_csv('manual_labeling.csv')
```

## 2- Checking Data Lengths and Comparing DataFrames

It's important to ensure that the DataFrames have the same length before comparing their contents.If the lengths match, we will compare the contents of the DataFrames row-by-row.

```python
if len(df1) != len(df2):
    print("The files have different lengths.")
else:
    for index, (row_df1, row_df2) in enumerate(zip(df1.iterrows(), df2.iterrows())):
        file1, cell1 = row_df1[1]['File'], row_df1[1]['Cell']
        file2, cell2 = row_df2[1]['File'], row_df2[1]['Cell']
        
        if file1 != file2 or cell1 != cell2:
            print(f"Difference detected at line {index + 1}:")
            print(f"File 1 - File: {file1}, Cell: {cell1}")
            print(f"File 2 - File: {file2}, Cell: {cell2}")
```
